```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load Data
sentiment =pd.read_csv("datasets/fear_greed_index.csv")
trades=pd.read_csv("datasets/historical_data.csv")
print("Sentiment shape: ",sentiment.shape)
print("Trades shape: ",trades.shape)
```

```python
print("Missing values in trades:\n", trades.isnull().sum())
print("Missing values in sentiment:\n", sentiment.isnull().sum())

print("Duplicate trades:", trades.duplicated().sum())
print("Duplicate sentiment rows:", sentiment.duplicated().sum())
```

```python
#Data preparation
#converting sentiment data coloumn
sentiment['date']=pd.to_datetime(sentiment['date'],errors='coerce')
sentiment['Date']=sentiment['date'].dt.date
#converting trades data coloumn
trades['Timestamp IST']=pd.to_datetime(trades['Timestamp IST'],dayfirst=True,errors='coerce')
trades['Date']=trades['Timestamp IST'].dt.date
```

```python
trades['Timestamp IST'].isnull().sum()
```

```python
#Merge Datasets
merged=trades.merge(sentiment[['Date','classification']],on='Date',how='left')
merged.head()
```

```python
print(sentiment.columns)
```

```python
merged['classification'].isnull().sum()
```

```python
merged.groupby('classification')['Closed PnL'].mean()
```

```python
#Trade count by sentiment
merged.groupby('classification').size()
```

```python
#Sentiment analysis
merged.groupby('classification')['Size USD'].mean()
```

```python
#While trading frequency increases during Fear periods, traders allocate significantly larger position sizes during Greed periods. This suggests increased risk-taking behavior when market sentiment is positive.
```

```python

#Combined Summary Table
summary = merged.groupby('classification').agg({
    'Closed PnL': 'mean',
    'Size USD': 'mean',
    'classification': 'count'
}).rename(columns={
    'Closed PnL': 'Avg_PnL',
    'Size USD': 'Avg_Size_USD',
    'classification': 'Trade_Count'
})
```

```python
summary
```

```python
summary.sort_values(by='Avg_PnL',ascending=False)
```

```python
# Visualization
#PnL distribution
sns.boxplot(x='classification', y='Closed PnL', data=merged)
plt.title("PnL Distribution by Sentiment")
plt.show()
```

```python
#Trades Size Comparison
sns.barplot(x=summary.index, y=summary['Avg_Size_USD'])
plt.xticks(rotation=45)
plt.title("Average Trade Size by Sentiment")
plt.show()
```

```python
#Average PnL by Sentiment
merged['Win']=merged['Closed PnL']>0
win_by_sentiment = merged.groupby('classification')['Win'].mean() * 100
win_by_sentiment
```

```python
#Win Rate by Sentiment


win_by_sentiment = merged.groupby('classification')['Win'].mean() * 100
win_by_sentiment
```

```python
#Drawdown Proxy

merged.groupby('classification')['Closed PnL'].std()
```

```python
#Trade Frequency

merged.groupby('classification').size()

# Position Size
merged.groupby('classification')['Size USD'].mean()


# Long / Short Bias
pd.crosstab(
    merged['classification'],
    merged['Side'],
    normalize='index'
) * 100


```

```python
#Frequent vs Infrequent Traders


trades_per_account = merged.groupby('Account').size()

median_trades = trades_per_account.median()

frequent_accounts = trades_per_account[trades_per_account > median_trades].index

merged['Frequency_Group'] = np.where(
    merged['Account'].isin(frequent_accounts),
    'Frequent',
    'Infrequent'
)

merged.groupby('Frequency_Group')['Closed PnL'].mean()
#Segment 3 — Consistent vs Inconsistent Traders



win_rate_acc = merged.groupby('Account')['Win'].mean()

consistent_accounts = win_rate_acc[win_rate_acc > 0.6].index

merged['Consistency_Group'] = np.where(
    merged['Account'].isin(consistent_accounts),
    'Consistent',
    'Inconsistent'
)

merged.groupby('Consistency_Group')['Closed PnL'].mean()
```

```python
#Trade Size by Sentiment
sns.barplot(x=summary.index, y=summary['Avg_Size_USD'])
plt.xticks(rotation=45)
plt.title("Average Trade Size by Sentiment")
plt.show()
```
